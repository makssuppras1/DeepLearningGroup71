## MNIST

This dataset defines a multiclass classification task, where each MNIST image must be categorized into one of ten digit classes (0–9). Each image has $28 \times 28 = 784$ pixels, with 8-bit grayscale values (0–255) representing the digit intensity. Pixels are stored in the input vector `X`. The output uses one-hot encoding: a vector `Y = (Y0, ..., Y9)` with a 1 for the digit class and 0s elsewhere. The dataset includes 60,000 training and 10,000 test images.


### Activation Function
The network has 784 units in the input layer and 10 units in the output layer. These outputs represent a single categorical variable in the digit class 0–9, meaning they are dependent: only one exclusive output is “active” for each image. To let the estimates represent class propability, the softmax activation function is applied in the output layer.

The predicted probability for class $c$ is:

$$
\hat{y}_c = p(y = c) = \frac{\exp(b_c + w_c^\top h^{(L)})}{\sum_{k=1}^{K} \exp(b_k + w_k^\top h^{(L)})}
$$

The predicted class is then chosen as:

$$
c_{\text{pred}} = \arg\max_c \hat{y}_c
$$


The softmax ensures the outputs are valid probabilities (non-negative and summing to one). The classifier assigns the image to the class with the highest probability.


## Loss Function


For this network, the target is categorical, so training minimizes the **cross-entropy loss**. The cross-entropy loss used for training is:

$$
\mathcal{L} = -\frac{1}{N} \sum_{j=1}^{N} \sum_{n=1}^{J} y_{j,c} \log(\hat{y}_{j,c})
$$

where $y_{j,c}$ is 1 if sample $j$ belongs to class $c$, and 0 otherwise.

Cross-entropy ensures the network outputs valid class probabilities and learns to assign the highest probability to the correct class.




## CIFAR-10

The CIFAR-10 dataset defines a multiclass classification task, where each image must be classified into one of ten object categories: airplane, automobile, bird, cat, deer, dog, frog, horse, ship, and truck. The pixel values for each image are organized in a three-dimensional array called a feature map. The first two axes correspond to the spatial dimensions (32 × 32), and the third axis represents the three color channels (RGB), giving a total of $32 \times 32 \times 3 = 3,072$ features per image. For fully connected networks, the feature map is flattened into a single vector and stored in the input vector $X$. The output uses one-hot encoding: a vector $Y = (Y_0, ..., Y_9)$ with a 1 for the correct class and 0s elsewhere. The dataset contains 50,000 training images and 10,000 test images.

### Hyperparameters & Network Capacity

The `width` of a neural network corresponds to the number of hidden units in each layer, while its `depth` reflects the number of hidden layers. The total number of hidden units indicates the overall capacity of the model.


Let $K$ denote the total number of layers, and $D_1, D_2, \dots, D_K$ the number of hidden units in each layer. These are `hyperparameters`, meaning they are set before estimating the model’s parameters (such as weights and biases). With fixed hyperparameters, the model defines a class of functions, and the learned parameters determine the specific function selected from this class.

A deep neural network with multiple layers can be described using the following notation. Let $\mathbf{h}_k$ denote the vector of hidden units at layer $k$, $\boldsymbol{\beta}_k$ the bias vector for the next layer, and $\boldsymbol{\Omega}_k$ the weight matrix connecting layer $k$ to layer $k+1$. The network maps an input $\mathbf{x}$ to an output $\mathbf{y}$ through successive transformations:

$$
\begin{aligned}
\mathbf{h}_1 &= a(\boldsymbol{\beta}_0 + \boldsymbol{\Omega}_0 \mathbf{x}) \\
\mathbf{h}_2 &= a(\boldsymbol{\beta}_1 + \boldsymbol{\Omega}_1 \mathbf{h}_1) \\
&\;\;\vdots \\
\mathbf{h}_K &= a(\boldsymbol{\beta}_{K-1} + \boldsymbol{\Omega}_{K-1} \mathbf{h}_{K-1}) \\
\end{aligned}
$$


where $a[\cdot]$ is the activation function applied at each layer.\\


The output then becomes:

$$
\begin{aligned}
\mathbf{y} &= \boldsymbol{\beta}_K + \boldsymbol{\Omega}_K \mathbf{h}_K
\end{aligned}
$$

or as one single function:  

$$
y = \boldsymbol{\beta}_K + \boldsymbol{\Omega}_K a \boldsymbol{\beta}_{K-1} + \boldsymbol{\Omega}_{K-1} a \Big[ \dots \boldsymbol{\beta}_1 + \boldsymbol{\Omega}_1 a (\boldsymbol{\beta}_0 + \boldsymbol{\Omega}_0 \mathbf{x}) \dots \Big]
$$

### Weights and Biases

If layer $k$ contains $D_k$ hidden units, the corresponding bias vector $\boldsymbol{\beta}_{k-1}$ has length $D_k$. The final bias vector $\boldsymbol{\beta}_K$ matches the output dimension $D_o$.  

The weight matrix connecting the input to the first hidden layer, $\boldsymbol{\Omega}_0$, has dimensions $D_1 \times D_i$, where $D_i$ is the input size. The last weight matrix, $\boldsymbol{\Omega}_K$, is of size $D_o \times D_K$, while the intermediate weight matrices $\boldsymbol{\Omega}_k$ have dimensions $D_{k+1} \times D_k$.


## Training Neural Networks

Training a neural network involves estimating the parameters to minimize a loss function, Eg. cross-entropy loss for classification tasks. Neural network loss functions are generally non-convex, so training does not guarantee convergence to a global optimum.

Training is performed using gradient-based optimization methods. Gradients of the loss function are computed with respect to the weights and backpropagated through the network using the chain rule. Weights are updated by taking a step of size $λ$ in the direction of the gradient:

### SDG and Mini-batches

In practice, training is often performed using mini-batches rather than the full dataset. A mini-batch provides a reasonably accurate, though noisy, estimate of the true gradient, which enables more frequent parameter updates and can help the optimizer escape poor local minima. For example, using batches of 128 observations allows for many updates per epoch while still approximating the full gradient effectively. ## Regularization


During training with **stochastic gradient descent (SGD)** and mini-batches, the noisy updates act as a form of implicit regularization. Using smaller mini-batches introduces variability in the gradient, which can prevent the model from overfitting to the training set. Other ways to reduce overfitting include constraining model capacity or modifying the training. 

#### Ridge Regularization (L2)

A common approach is weight decay (L2 regularization), which penalizes large weights by adding a quadratic term to the loss and thereby encourages smoother functions. 

$$
\mathcal{L}_{\text{total}} = \mathcal{L}_{\text{data}} + \lambda \sum_j \theta_j^2
$$

where:
- $\mathcal{L}_{\text{data}}$ is the original loss function,  
- $\theta_j$ are the model parameters, and  
- $\lambda$ controls the strength of regularization.  

Smaller $\lambda$ reduces the effect, while larger $\lambda$ enforces stronger weight shrinkage. Different layers can use different $\lambda$ values if desired.  


#### Lasso Regularization (L1)

L1 regularization is another variant which pushes many weights exactly to zero and enforces sparsity. Instead of adding a penalty proportional to the sum of squared weights like L2, it adds a penalty proportional to the sum of the absolute values of the weights

$$
\mathcal{L}_{\text{total}} = \mathcal{L}_{\text{data}} + \lambda \sum_j |\theta_j|
$$

L1 regularization encourages sparsity, pushing some weights exactly to zero, which can help with feature selection or pruning.

#### Early stopping

Early stopping leverages the validation set to detect overfitting during training. The validation objective is monitored across epochs, and training is halted once this objective begins to increase. This prevents the model from fitting noise in the training data and functions as an effective regularization mechanism.

#### Dropout 

Dropout is a regularization technique to prevent overfitting. During training, a random fraction of neurons is deactivated at each step, so the number of active nodes and their connectivity vary over time. Each weight update is computed with a different subset of the network, as training multiple networks in parallel. Dropout reduces overfitting and improves generalization. During inference or testing, dropout is disabled and all neurons remain active.

## Tuning the Network


When training a neural network, several choices influence performance. One important factor is the architecture of the network, specifically the **number of hidden layers** and the **number of units** in each layer. Here, risk of overfitting can be mitigated through appropriate regularization techniques.

Howwever, regularization requires careful tuning. Parameters such as the **dropout rate** and the strength of **Lasso- or Ridge penalties (λ)** can be set independently for each layer. Dropout randomly deactivates units during training, preventing nodes from becoming over-specialized, while L1 and L2 penalties constrain the magnitude of the weights, encouraging sparsity or small values, respectively.

Stochastic gradient descent (SGD) is the standard method for optimizing neural networks. Hyperparameters in SGD include the **batch size**, which determines how many training samples are used for each gradient update, and the **number of epochs**, which controls how many passes over the entire dataset are performed.
